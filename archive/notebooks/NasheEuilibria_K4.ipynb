```python
# Retry analysis with itertools now properly imported

import pickle
import itertools
from collections import defaultdict
import numpy as np

# Load simulation result
with open("/projects/esthercui/ewl_k4_parallel_results1.pkl", "rb") as f:
    data = pickle.load(f)

k = 4  # number of players
GRID_THETA = 11
GRID_PHI = 6
GRID_LEN = GRID_THETA * GRID_PHI

# Map profile index to (θ, φ)
theta_vals = np.linspace(0, np.pi, GRID_THETA)
phi_vals = np.linspace(0, np.pi/2, GRID_PHI)
GRID = list(itertools.product(theta_vals, phi_vals))
param_lookup = {idx: vals for idx, vals in enumerate(GRID)}

# Build payoff map
payoff_map = {profile: payoff for profile, payoff in data}

# Best-response mapping
best_resp_value = defaultdict(lambda: -np.inf)
best_resp_choice = {}

for profile, payoff in payoff_map.items():
    for p in range(k):
        opp_profile = profile[:p] + profile[p+1:]
        key = (p, opp_profile)
        if payoff[p] > best_resp_value[key]:
            best_resp_value[key] = payoff[p]
            best_resp_choice[key] = profile[p]

# Find Nash equilibria
nash_profiles = []
for profile, payoff in payoff_map.items():
    if all(profile[p] == best_resp_choice[(p, profile[:p]+profile[p+1:])] for p in range(k)):
        strategy_angles = tuple(param_lookup[i] for i in profile)
        nash_profiles.append((strategy_angles, payoff))

nash_profiles[:5], len(nash_profiles)
```
